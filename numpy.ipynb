```python
x=np.random.randn(5)
y=np.random.randn(5)
z=np.random.randn(5)
print(x)
print(y)
print(z)
```

```python
print(np.minimum(np.minimum(x,y),z))      # 두 요소를 비교하여 작은값 찾기 (처음 요소와 마지막 요소를 비교)
print(np.maximum(np.maximum(x,y),z))      # 두 요소를 비교하여 큰값 찾기 (처음 요소와 마지막 요소를 비교)
```

정렬

```python
arr=np.array([[3,2,4],[2,7,6],[9,5,1]])
print(arr)
print(np.sort(arr))             # 행기준 정렬
print(np.sort(arr, axis=0))     # 열기준 정렬
print(np.sort(arr, axis=1))     # 행기준 정렬
print(np.sort(arr, axis=0)[::-1,:])    # 열기준 행의 순서를 역순으로 정렬
print(np.sort(arr, axis=1)[:,::-1])    # 행기준 열의 순서를 역순으로 정렬
```

사용자별 영화평점의 평균                  유저아이디별 영화평점의 평균

```python
user_ids=np.unique(data[:,0])
print(len(user_ids))
mean_values=[]
for user_id in user_ids:
    data_user=data[data[:,0]==user_id]
    value=data_user[:,2].mean()
    mean_values.append([user_id,value])

mean_values[:5]
```

```python
print(len(mean_values))
mean_array=np.array(mean_values)
print(mean_array.shape)
np.savetxt("data/result.csv",mean_array,fmt="%.1f", delimiter=",")
```
