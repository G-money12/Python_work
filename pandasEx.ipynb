pd.DataFrame(data, columns=[], index=[])

```python
data={
    "name":['민준','현우','서연','동현','지현'],        # 하나가 필드
    "year":[2023,2024,2022,2025,2024],                # 하나가 필드
    "point":[1.5,1.7,3.6,2.4,2.9]                     # 하나가 필드
    }
df=pd.DataFrame(data)
df
```

```python
names=['민준','현우','서연','동현','지현']
years=[2023,2024,2022,2025,2024]
points=[1.5,1.7,3.6,2.4,2.9]

data=[names, years, points]
print(data)
df1=pd.DataFrame(data).T            # .T 데이터를 가져왔는데 행과 열을 바꾸고 싶을때 .T
df1.columns=['names','years','points'];df1      # python 명령어를 두개 적을때 ; 새미콜론 기입한다.
```

```python
arr=np.array(data).T
arr
df2=pd.DataFrame(arr,columns=["names","years","points"],        # columns 가로 바
                 index=['one','two','theree','four','five'])    # index 세로 바
df2
```

```python
print(df2)
print(df2.index)
print(df2.columns)
print(df2.values)
```

```python
names=df2['names']
years=df2['years']
points=df2['points']
print(names)
print(type(years))
print(points)
```
