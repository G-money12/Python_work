```python
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
```

Graph

```python
def chart1():
    x=[1,2,3,4,5]
    y=[10,20,25,30,40]

    plt.figure(figsize=(5,4),dpi=128)       # dpi 인치당 128도트
    
    plt.plot(x,y)
    plt.title("Simple Line Graph")
    plt.xlabel("X Axis")
    plt.ylabel("Y Axis")
    plt.savefig("data/chart1.png")
    plt.show()
    
chart1()
```

```python
x=[1,2,3,4,5]
y1=[10,20,25,30,40]
y2=[5,15,20,25,30]

plt.plot(x,y1, label='Line 1', color='blue', linestyle='--', marker='o')
plt.plot(x,y2, label='Line 2', color='red', linestyle='-', marker='s')
plt.title("Multipe Line Graphs")
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.legend()
plt.grid(True)
plt.savefig("data/mult_chart.png")
plt.show()
```

Bar Chart

```python
categories=['A','B','C','D']
values=[10,20,15,25]
plt.bar(categories, values, color=['skyblue','red','green','purple'])
```

```python
categories=['A','B','C','D']
values1=[10,20,15,25]
values2=[15,12,20,15]
plt.bar(categories, values1, color=['skyblue','orange','pink','yellow'])
plt.bar(categories, values2, color=['blue','red','green','purple'], bottom=values1)
```

Histogram Chart

```python
data = np.random.randn(1000)        # 정규 분포를 따르는 난수 생성
plt.hist(data, bins=10, color='blue', alpha=0.7)        # bins는 막대 개수, alpha는 불투명도

plt.title("Histogram Example")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

Scatter (산점도) 차트

```python
x= np.random.rand(50)

y= np.random.rand(50)

colors=np.random.rand(50)
sizes=np.random.rand(50)*500
plt.scatter(x,y, c=colors, s=sizes, alpha=0.5, cmap='viridis')

plt.title("Military by age")
plt.xlabel("Years")
plt.ylabel("Gender")

plt.colorbar()
plt.show()
```

Pie(파이) Chart

```python
labels=['A','B','C','D']
sizes=[10,20,30,40]
colors=['blue','green','red','purple']
explode=(0,0.1,0,0)     # 두 번째 조각 강조

plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', explode=explode, startangle=90)

plt.title("Pie Chart Example")

plt.show()
```

Subplots (서브플롯)

```python
import matplotlib.font_manager as fm       # 한글 나타내기
plt.rc('font', family='Malgun Gothic')
plt.rc('axes', unicode_minus=False)         # 한글 서체로 바꾸면 - 도 깨져서 나옴, 그래서 유니코드로 -만들는것을 반대한다.

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

plt.figure(figsize=(10, 5))     # 그래프 크기 설정 (인치)

# 첫 번째 그래프 (사인 함수)
plt.subplot(2, 1, 1)        # (행, 열, 위치) 
plt.plot(x, y1, label="Sin", color='red')
plt.title("싸인 함수")
plt.legend()        # 범례

# 두 번째 그래프(코사인 함수)
plt.subplot(2, 1, 2)
plt.plot(x, y2, label="Cos", color='blue')
plt.title("코사인 함수")
plt.legend()

# plt.subplot(1, 2, 1)
# plt.plot(x, y3, label="T", color='green')
# plt.title("T Function")
# plt.legend()



plt.tight_layout()      # 그래프 간격 조정

plt.show()
```

temperature_2014

```python
data=pd.read_csv("data/temperature_2014.csv")
data.info()
```

```python
data['PST']=pd.to_datetime(data['PST'])
data.info()
```

```python
plt.rc('font', family='Malgun Gothic')      # 한글처리시 
plt.rc('axes', unicode_minus=False)         # 세트 메뉴
fig=plt.figure(dpi=128, figsize=(10,6))
pst=list(data['PST'])                       # 요렇게 해줘야 list형식
high=list(data['Max TemperatureF'])
low=list(data['Min TemperatureF'])
plt.plot(pst, low, c='blue', alpha=0.5)
plt.plot(pst, high, c='red', alpha=0.5)
plt.fill_between(pst, high, low, facecolor='skyblue', alpha=0.5)

plt.title("온도 통계", fontsize=20)
plt.xlabel("기간", fontsize=16)
plt.ylabel("온도", fontsize=16)
plt.tick_params(axis='both', labelsize=12)
plt.savefig("data/temp_chart.png")
plt.show()                  # 웹에서는 show를 하지않으면 나오지 않는다.
```
