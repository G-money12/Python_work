# 1. QLite (어플리케이션용 DB)

가벼운 데이터 베이스 , 앱안에 파일형태로 DB 사용 , 작은 메모리 사용시

python 만 설치해도 기본으로 들어있다.

```python
import sqlite3
```

table 생성

```python
def create_table():
    conn = sqlite3.connect('booksdb')                               # 데이터베이스 커넥션 생성
    cur = conn.cursor()                                             # 쿼리 명령을 위한 커서 생성
                                                                    # 테이블 생성(제목, 출판일자, 출판사, 페이지수, 추천영부)
    sql="""CREATE TABLE IF NOT EXISTS books (
                                            title TEXT,        
                                            publicshed_date TEXT, 
                                            publisher TEXT, 
                                            page INTEGER, 
                                            recommend INTEGER
                                            )"""    
    cur.execute(sql)    # 쿼리 명령 수행   
    conn.commit()       # 데이터베이스 변경내용 완료
    conn.close()        # 커넥션 닫기 
create_table()
```

데이터 추가

```python
def insert_book1():
    conn = sqlite3.connect('booksdb')
    cursor=conn.cursor()
    sql="INSERT INTO books VALUES ('Java','2019-05-20','길벗',500,10)"
    cursor.execute(sql)
    conn.commit()
    conn.close()

insert_book1()             # 함수를 벚어나서 호출해야함
```

데이터 조회(모든 데이터 조회)

```python
def select_all():
    conn = sqlite3.connect('booksdb')
    cursor=conn.cursor()
    sql="SELECT * FROM books"
    cursor.execute(sql)         # cursor.execute("select * from books")
    books=cursor.fetchall()     # 실행된 SQL 쿼리의 결과를 모두 가져오는 메서드 , 한 행의 데이터를 튜플로 가져오며, 전체 결과는 list에 저장
    for book in books:
        print(books)
    conn.close()
select_all()
```

데이터 1개 넣고 전체 조회

```python
def insert_book2(data):
    conn = sqlite3.connect('booksdb')
    cursor=conn.cursor()
    #sql="INSERT INTO books VALUES (?,?,?,?,?)"
    sql="INSERT INTO books VALUES (?,?,?,?,?)"
    cursor.execute(sql,data)       # sql , 튜플형태로 데이터 넣기
    conn.commit()
    conn.close()

data = ('Python','2019-05-20','한빛',584,20)
insert_book2(data)
select_all()
```

복수행의 데이터 추가

```python
def many_insert_book(datas):
    conn=sqlite3.connect('booksdb')
    cursor=conn.cursor()
    sql="INSERT INTO books VALUES(?,?,?,?,?)"
    cursor.executemany(sql,datas)
# for data in datas:
#     cursor.execute(sql,data) 
    conn.commit()
    conn.close()

datas=[('Python','2019-05-20','두빛',584,20),   
        ('Java','2019-05-20','두벗',500,10),
        ('JavaScript','2019-05-20','두빛',584,20)]
many_insert_book(datas)
select_all()
```

데이터 조회

쿼리 결과에서 1개의 데이터 출력

```python
def one_book():
    conn=sqlite3.connect('booksdb')
    cursor=conn.cursor()
    cursor.execute("SELECT * FROM books")                             # 쿼리 결과에서 한개의 데이터를 가져옴
    book=cursor.fetchone()
    print(book)
    conn.close()
one_book()
```

쿼리 결과에서 지정된 갯수의 데이터 출력

```python
def some_book(number):
    conn=sqlite3.connect('booksdb')
    cursor=conn.cursor()
    cursor.execute("SELECT * FROM books")                             # 쿼리 결과에서 두개의 데이터를 가져옴
    book=cursor.fetchmany(number)
    # for book in books:
    #     pring(book)
    print(book)
    conn.close()
some_book(2)
```

데이터 조회(조건 검색)

```python
def search_book(page, recommend):
    conn=sqlite3.connect('booksdb')
    cur=conn.cursor()
    sql="SELECT * FROM books WHERE page>=? and recommend>=? order by title desc"
    cur.execute(sql, (page, recommend))
    books=cur.fetchall()
    for book in books:
        print(book)
    conn.close()

search_book(500, 5)
```

update      

update정보가 1개일때

```python
def update_book(data):
    conn=sqlite3.connect('booksdb')
    cur=conn.cursor()
    sql="UPDATE books SET page=?, recommend=? WHERE title=?"
    cur.execute(sql,data)
    conn.commit()
    conn.close()

data=(500, 8, 'Java')
update_book(data)
select_all()
    
```

update 정보가 복수일 때

```python
def update_books(data):
    conn=sqlite3.connect('booksdb')
    cur=conn.cursor()
    sql="UPDATE books SET page=?, recommend=? WHERE title=? "
    cur.executemany(sql,data)
    conn.commit()
    conn.close()

datas=[(100, 1, 'Java'), (200, 2, 'Python')]
update_books(datas) 
select_all()
```

데이터 삭제

삭제 조건이 단수일때

```python
def delete_book(data):
    conn=sqlite3.connect('booksdb')
    cur=conn.cursor()
    sql="DELETE FROM books WHERE title=?"
    cur.execute(sql,(data,))
    conn.commit()
    conn.close() 

delete_book('Java')   
select_all()
```

삭제 조건이 복수일 때

```python
def delete_books(datas):
    conn=sqlite3.connect('booksdb')
    cur=conn.cursor()       
    sql="DELETE FROM books WHERE title=?"
    cur.executemany(sql,datas)              # executemany : 조건이 여러개일때  , execute : 조건이 하나일때 (기준이 조건이 하나의 튜플이면 execute이다.)
    conn.commit()
    conn.close()

datas=[('Python',), ('JavaScript',)]
delete_books(datas)
select_all()
```
